```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df =pd.DataFrame({
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Karan", "Neha", "Arjun", "Sneha", "Vikas", "Anjali"],
    "Math": [85, 92, 78, 88, 75, 95, 82, 90, 70, 87],
    "Python": [90, 95, 80, 85, 72, 98, 88, 91, 76, 89],
    "English": [78, 88, 75, 90, 80, 92, 85, 87, 72, 84],
})
```

```python
plt.plot(df["Name"], df["Math"], label= "Math Marks", marker="^")
plt.plot(df["Name"], df["Python"], label= "Python Marks", marker="o")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.title("Math and Python Marks Analysis")
plt.grid()
plt.savefig("Math and Python Marks Analysis.png")
```

```python
plt.boxplot([
    df["Math"],
    df["Python"],
    df["English"]
])

plt.xticks([1, 2, 3], ["Math", "Python", "English"])
plt.ylabel("Marks")
plt.title("Marks Distribution")
plt.grid()

plt.show()
```
