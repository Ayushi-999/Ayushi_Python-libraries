```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.DataFrame({
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Vikas",
             "Anjali", "Rohit", "Sneha", "Karan", "Pooja"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19, 22, 20],
    "Python": [85, 78, 92, 88, 67, 95, 72, 89, 80, 91],
    "Java": [80, 75, 90, 85, 70, 92, 68, 86, 78, 94],
    "Math": [88, 82, 95, 79, 65, 90, 74, 91, 83, 87],
    "Gender": ["Male", "Male", "Female", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Department": ["BCA", "BCA", "BSc", "BCA", "BSc",
                   "BCA", "BSc", "BCA", "BSc", "BCA"]
})
df
```

```python
sns.barplot(data=df, x="Name", y="Python", hue="Gender")
sns.barplot(data=df, x="Name", y="Java", hue="Gender")
plt.xticks(rotation=75)
plt.show()
```

```python
sns.countplot(data=df, x="Department", hue="Gender")
```

```python
ax= sns.countplot(data=df, x="Department", hue="Gender")
for i in ax.containers:
    ax.bar_label(i)
```

```python
corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```
